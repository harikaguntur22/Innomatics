```python
# Problem 1: Mobile Recharge Validation System

def validate_recharge():
    valid_plans = [199, 299, 399, 599]

    while True:
        amount = int(input("Enter recharge amount: ₹"))

        if amount < 50:
            print("Recharge amount must be at least ₹50.")
        elif amount not in valid_plans:
            print("Invalid plan. Available plans:", valid_plans)
        else:
            print("Recharge Successful!")
            break

validate_recharge()


# Problem 2: Inventory Reorder Alert System

def check_inventory(stock):
    for product, quantity in stock.items():
        if quantity < 15:
            print(product, "- Reorder Alert")
        else:
            print(product, "- Stock OK")

products = {
    "Laptop": 20,
    "Mouse": 10,
    "Keyboard": 12,
    "Monitor": 18
}

check_inventory(products)


# Problem 3: Student Result Processing System

def process_result(marks):
    total = 0

    for mark in marks:
        total += mark

    average = total / len(marks)

    print("Average Marks:", average)

    if average >= 50:
        return "Pass"
    else:
        return "Fail"

student_marks = [60, 70, 55, 45, 65]

result = process_result(student_marks)
print("Result:", result)


# Problem 4: Cab Fare Estimator with Retry Option

def calculate_fare(distance, peak_hour):
    fare = 50 + (distance * 12)

    if peak_hour.lower() == "yes":
        fare += fare * 0.25

    return fare

while True:
    distance = float(input("Enter distance (km): "))
    peak = input("Peak hour? (yes/no): ")

    fare = calculate_fare(distance, peak)

    print("Estimated Fare: ₹", fare)

    retry = input("Do you want to calculate again? (yes/no): ")

    if retry.lower() != "yes":
        break


# Problem 5: Employee Attendance Eligibility Checker

def attendance_eligibility(attendance):
    present_days = 0

    for day in attendance:
        if day == "P":
            present_days += 1

    percentage = (present_days / len(attendance)) * 100

    print("Attendance Percentage:", percentage)

    if percentage >= 75:
        return "Eligible"
    else:
        return "Not Eligible"

attendance_list = ["P", "P", "A", "P", "P", "A", "P", "P"]

result = attendance_eligibility(attendance_list)
print(result)


# Problem 6: Password Strength Checker

def check_password(password):
    special_chars = "@#$"

    has_digit = False
    has_special = False

    if len(password) < 8:
        return "Weak Password"

    for char in password:
        if char.isdigit():
            has_digit = True

        if char in special_chars:
            has_special = True

    if has_digit and has_special:
        return "Strong Password"
    else:
        return "Weak Password"

password = input("Enter Password: ")
print(check_password(password))
```
