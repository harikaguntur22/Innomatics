```python
#Problem 1: Smart Parking Lot Management System
def parking_management(capacity, logs):
    parked = 0
    peak = 0

    for log in logs:
        if log.upper() == "IN":
            parked += 1
        elif log.upper() == "OUT" and parked > 0:
            parked -= 1

        if parked > peak:
            peak = parked

    print("Currently Parked Vehicles:", parked)
    print("Peak Parking Usage:", peak)

    if parked > capacity:
        print("Parking Status: Full")
        print("Alert: Parking Capacity Exceeded")
    else:
        print("Parking Status: Available")


capacity = 50
logs = ["IN", "IN", "IN", "OUT", "IN", "IN", "OUT"]

parking_management(capacity, logs)
#Problem 2: Online Food Delivery Time Estimator
def delivery_time(distance, traffic, weather):
    eta = distance * 5

    if traffic.lower() == "medium":
        eta += 10
    elif traffic.lower() == "high":
        eta += 15

    if weather.lower() == "rainy":
        eta += 10
    elif weather.lower() == "storm":
        eta += 20

    print("Estimated Delivery Time:", eta, "minutes")


delivery_time(8, "High", "Rainy")
#Problem 3: Movie Theatre Seat Occupancy Analyzer
def seat_occupancy(total_seats, booked_seats):
    booked = len(booked_seats)
    occupancy = (booked / total_seats) * 100

    if occupancy == 100:
        status = "Housefull"
    elif occupancy >= 75:
        status = "Almost Full"
    else:
        status = "Seats Available"

    print("Occupancy:", occupancy, "%")
    print("Show Status:", status)

    if occupancy >= 90:
        print("Suggestion: Open Additional Show")


booked = [1] * 150

seat_occupancy(200, booked)
#Problem 4: Cloud Server Load Classification System
def server_load(cpu_readings):
    total = 0

    for reading in cpu_readings:
        total += reading

    average = total / len(cpu_readings)

    if average < 50:
        status = "Normal"
    elif average <= 80:
        status = "Warning"
    else:
        status = "Critical"

    print("Average CPU Load:", average)
    print("Server Status:", status)


cpu = [45, 60, 70, 85, 90]

server_load(cpu)
#Problem 5: Smart Classroom Resource Usage Monitor
def resource_monitor(resources):
    overused = []

    for resource, hours in resources.items():
        if hours > 8:
            overused.append(resource)

    if overused:
        print("Overused Resources:", ", ".join(overused))
        print("Energy Alert: Yes")
    else:
        print("Overused Resources: None")
        print("Energy Alert: No")


usage = {
    "Projector": 6,
    "AC": 9,
    "Lights": 4
}

resource_monitor(usage)
#Problem 6: Online Event Registration Capacity Controller
def event_registration(capacity, registrations):
    confirmed = 0

    for _ in range(min(capacity, registrations)):
        confirmed += 1

    waitlisted = max(0, registrations - capacity)

    if registrations >= capacity:
        status = "Closed"
    else:
        status = "Open"

    print("Confirmed Registrations:", confirmed)
    print("Waitlisted Users:", waitlisted)
    print("Registration Status:", status)


event_registration(100, 105)
```
